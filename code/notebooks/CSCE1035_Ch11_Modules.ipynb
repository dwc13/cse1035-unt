# Chapter 11: Modules

## What is a Module?
A module is a file containing Python definitions and statements (ending with the .py extension).

**Why Use Modules?**

Allowing you to organize your code into separate files:
- Easier to understand
- Improves organization; separates functionality classes, functions, variables
- Allows for reusability

## Example Modules Section

These module files will be created once and used throughout the notebook.

**DO NOT EDIT THESE! JUST RUN**

````python
%%writefile greetings.py
def say_hello(name):
    print(f"Hello, {name}!")

print(f"Module name is: {__name__}")

if __name__ == "__main__":
    say_hello("World")
````

````python
%%writefile math_utils.py
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

print(f"Module name is: {__name__}")

if __name__ == "__main__":
    print("Running math_utils directly")
    print("2 + 3 =", add(2, 3))
else:
    print("Importing math_utils")
````

# Reloading Modules
To reload a module after editing, use the function `reload()` from the package `importlib`.

Note:
May not work in some interactive environments like Jupyter Notebooks without additional steps.
- Restart the kernel if issues arise.
    - Google Colab disconnect and reconnect the runtime to reset the environment. 
    - VSCode Jupyter Notebooks restart the kernel at the top of the notebook.

## Example
Import `math_utils` and use its functions before reloading. Once an import is done, the module is cached, and further imports do not reflect changes made to the module file unless reloaded.

Then let's make some changes to `math_utils.py` and attempt to use functions again before the reload.

````python
import math_utils
import math_utils
import math_utils
print(math_utils.add(2, 3))
````

````python
%%writefile math_utils.py
def add(a, b):
    return 2*(a + b)

def sub(a, b):
    return a - b

print(f"Module name is: {__name__}")

if __name__ == "__main__":
    print("Running math_utils directly")
    print("2 + 3 =", add(2, 3))
else:
    print("Importing math_utils")
````

````python
print(math_utils.add(2, 3))
````

````python
import math_utils
print(math_utils.add(2, 3))
````

Now let's reload the module to see the changes take effect.

````python
import importlib
importlib.reload(math_utils)
print(math_utils.add(2, 3))
````
